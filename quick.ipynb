```python
import torch
import numpy as np
import matplotlib.pyplot as plt
```

```python
def print_results(results, name):
    mean = np.mean(results)
    std = np.std(results)
    ci = 1.96 * std / np.sqrt(len(results))
    print(f"{name}: {mean:.3f} ± {ci:.3f}")
```

```python
anchors = [2,3,4,5]
ours = [-6.093, -5.83, -5.76, -5.44]
eucl_w_potential = [-6.397, -5.97, -5.82, -5.44]
learned_no_potential = [-6.158, -5.93, -5.69, -5.44]
eucl_no_potential = [-6.307, -5.93, -5.73, -5.44]
cfm = [-6.251, -5.72, -5.75, -5.44]
direct = [-6.711, -5.79, -5.98, -5.44]

std_err = 0.03

plt.plot(anchors, ours, label="Ours", marker='o')
plt.fill_between(anchors, np.array(ours) - std_err, np.array(ours) + std_err, alpha=0.2)

plt.plot(anchors, eucl_w_potential, label="Eucl w potential", marker='o')
plt.fill_between(anchors, np.array(eucl_w_potential) - std_err, np.array(eucl_w_potential) + std_err, alpha=0.2)

plt.plot(anchors, learned_no_potential, label="Learned no potential", marker='o')
plt.fill_between(anchors, np.array(learned_no_potential) - std_err, np.array(learned_no_potential) + std_err, alpha=0.2)

plt.plot(anchors, eucl_no_potential, label="Eucl no potential", marker='o')
plt.fill_between(anchors, np.array(eucl_no_potential) - std_err, np.array(eucl_no_potential) + std_err, alpha=0.2)

plt.plot(anchors, direct, label="Direct", marker='o')
plt.fill_between(anchors, np.array(direct) - std_err, np.array(direct) + std_err, alpha=0.2)

plt.plot(anchors, cfm, label="CFM", marker='o')
plt.fill_between(anchors, np.array(cfm) - std_err, np.array(cfm) + std_err, alpha=0.2)


plt.xticks([2,3,4,5])
plt.show()
```
