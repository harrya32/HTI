```python
import torch
import numpy as np
import matplotlib.pyplot as plt
```

```python
def print_results(results, name):
    mean = np.mean(results)
    std = np.std(results)
    ci = std / np.sqrt(len(results))
    print(f"{name}: {mean:.3f} ± {ci:.3f}")
```
